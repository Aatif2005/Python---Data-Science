```python
import pandas as pd
df = pd.read_excel(r"C:\Users\aatif\OneDrive\Desktop\Instagram_Dataset_100_Rows.xlsx")
# first 5 rows ke liye
print("HEAD")
print(df.head())

# Information about columns, data types and non-null values
print("\nINFO")
df.info()

# Statistical summary of numerical columns
print("\nDESCRIBE")
print(df.describe())

# Dataset dimensions
print("\nSHAPE")
print(df.shape)

# Missing values in each column
print("\nMISSING VALUES")
print(df.isnull().sum())
```

```python
total = df["Number of Followers"].sum()
n = len(df["Number of Followers"])
mean = total / n
print("Sum =", total)
print("Number of observations =", n)
print("Mean =", mean)
```

```python
data = df["Number of Followers"].dropna()
data = sorted(data)
n = len(data)
if n % 2 == 1:
    median = data[n // 2]
else:
     median = (data[n // 2 - 1] + data[n // 2]) / 2
print("Median =", median)
```

```python
# standard deviation
import pandas as pd
import math
df = pd.read_excel(r"C:\Users\aatif\OneDrive\Desktop\Instagram_Dataset_100_Rows.xlsx")
data = df["Number of Followers"].dropna()
n = len(data)
mean = data.sum() / n
squared_differences = [(x - mean) ** 2 for x in data]
variance = sum(squared_differences) / n
standard_deviation = math.sqrt(variance)
print("Mean =", mean)
print("Variance =", variance)
print("Standard Deviation =", standard_deviation)
```

```python
#Experiment 1: Conduct basic data exploration by calculating summary statistics, creating histograms, and generating #scatterplots
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_excel(r"C:\Users\aatif\OneDrive\Desktop\Instagram_Dataset_100_Rows.xlsx")
print(df.head())
```

```python
print("Shape of dataset:", df.shape)
print("\nColumn names:")
print(df.columns)
print("\nData types:")
print(df.dtypes)
```

```python
# Select numerical columns
numeric_data = df.select_dtypes(include='number')
# Mean
print("Mean:")
print(numeric_data.mean())
# Median
print("\nMedian:")
print(numeric_data.median())
# Standard Deviation
print("\nStandard Deviation:")
print(numeric_data.std())
# Minimum
print("\nMinimum:")
print(numeric_data.min())
# Maximum
print("\nMaximum:")
print(numeric_data.max())
```
