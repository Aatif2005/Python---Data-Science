```python
# Yes. Restricted sample space means we take the original sample space and apply a condition, so only certain outcomes #remain possible.
#S(r) = {x∈S:Followers>1000}
restricted_sample = df[df["Number of Followers"]<50]
print("Restricted sample space size",len(restricted_sample))
print(restricted_sample)
```

```python
'''A discrete random variable is a variable whose possible values can be counted.
For your dataset, good examples would be:
Number of posts
Number of likes
Number of followers
Number of self-presenting posts
Number of comments, if available
For example, let:
X=Number of posts made by a randomly selected Instagram user'''
# Discrete Random Variable
X = df["Number of Posts"]

# Frequency
frequency = X.value_counts().sort_index()

# Probability
probability = frequency / len(X)

# Create probability distribution table
distribution = pd.DataFrame({
    "Number of Posts (X)": frequency.index,
    "Frequency": frequency.values,
    "Probability P(X)": probability.values
})

print(distribution)
```

```python
# Continuous Random Variable
# X = Engagement Rate

X = (df["Number of Likes for the Tenth Previous Post"] /
     df["Number of Followers"]) * 100

print("Engagement Rate:")
print(X)

print("Minimum:", X.min())
print("Maximum:", X.max())
print("Mean:", X.mean())
```

```python
X = df["Number of Posts"]
Y = df["Number of Followings"]
correlation = X.corr(Y)
print("Correlation between Number of Posts and Number of Followings:", correlation)
```

```python
# Dependent Variable
# X = Number of Posts
# Y = Number of Followers
X = df["Number of Posts"]
Y = df["Number of Followers"]
# Calculate correlation
correlation = X.corr(Y)
print("Correlation:", correlation)
```

```python
# Conditional Probability
A = df["Number of Followers"] > 1000
B = df["Number of Posts"] > 10
# Count B
count_B = B.sum()
# Count A and B together
count_A_and_B = (A & B).sum()
# Conditional Probability
P_A_given_B = count_A_and_B / count_B
print("P(A|B) =", P_A_given_B)
```
