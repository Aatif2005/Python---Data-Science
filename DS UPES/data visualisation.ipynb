```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
# Load the dataset
df = pd.read_csv(r"C:\Users\aatif\OneDrive\Desktop\cardekho_dataset.csv")

# Display first 5 records
print(df.head())

# Display dataset information
print(df.info())

# Display number of rows and columns
print("Dataset Shape:", df.shape)

# Check missing values
print(df.isnull().sum())
```

```python
# Count cars by brand
brand_count = df['brand'].value_counts().head(10)

# Create bar chart
plt.figure(figsize=(10, 5))

plt.bar(brand_count.index, brand_count.values)

plt.title("Top 10 Car Brands")
plt.xlabel("Car Brands")
plt.ylabel("Number of Cars")

plt.xticks(rotation=45)

plt.show()
```

```python
# Calculate average selling price for each vehicle age
age_price = df.groupby('vehicle_age')['selling_price'].mean()

# Create line graph
plt.figure(figsize=(10, 5))

plt.plot(age_price.index, age_price.values,
         marker='o', linestyle='-')

plt.title("Vehicle Age vs Average Selling Price")
plt.xlabel("Vehicle Age (Years)")
plt.ylabel("Average Selling Price")

plt.grid(True)

plt.show()
```

```python
# Select numerical columns
numeric_data = df.select_dtypes(include=['int64', 'float64'])

# Calculate correlation
correlation = numeric_data.corr()

# Create heatmap
plt.figure(figsize=(12, 8))

sns.heatmap(correlation,
            annot=True,
            cmap='coolwarm',
            fmt='.2f')

plt.title("Correlation Heatmap of Car Dataset")

plt.show()
```

```python
# Create box plot
plt.figure(figsize=(10, 6))

sns.boxplot(x='fuel_type',
            y='selling_price',
            data=df)

plt.title("Selling Price Distribution by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Selling Price")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(data=df,
             x='selling_price',
             bins=30,
             kde=True)

plt.title("Distribution of Selling Prices")
plt.xlabel("Selling Price")
plt.ylabel("Frequency")

plt.show()
```
