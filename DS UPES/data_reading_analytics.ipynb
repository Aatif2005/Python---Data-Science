```python
import pandas as pd
pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.width", None)
df = pd.read_excel(r"C:\Users\aatif\OneDrive\Desktop\Instagram_Dataset_100_Rows.xlsx")
print(df)
```

```python
print("First 5 rows:")
display(df.head())
```

```python
print("Shape of dataset ",df.shape)
```

```python
print("\nColumn names:")
print(df.columns.tolist())

# Data types and information
print("\nDataset information:")
df.info()

# Missing values
print("\nMissing values:")
print(df.isnull().sum())

# Duplicate rows
print("\nNumber of duplicate rows:")
print(df.duplicated().sum())
```

```python
# Make a copy so the original dataframe remains unchanged
df_clean = df.copy()

# Remove extra spaces from column names
df_clean.columns = df_clean.columns.str.strip()

# Display cleaned column names
print("Cleaned column names:")
print(df_clean.columns.tolist())
```

```python
# Check unique gender values
print("Unique Gender values:")
print(df_clean["Gender"].unique())

# Count each gender
print("\nGender counts:")
print(df_clean["Gender"].value_counts())
```

```python
# Numerical columns
numeric_columns = [
    "Number of Posts",
    "Number of Followers",
    "Number of Followings",
    "Number of Likes for the Tenth Previous Post",
    "Number of Likes for the Eleventh Previous Post",
    "Number of Likes for the Twelfth Previous Post",
    "Number of Self-Presenting Posts from Nine Previous Posts"
]

# Check for negative values
print("Negative values in each numerical column:")
print((df_clean[numeric_columns] < 0).sum())
# all the values are 0 that means no negative values
```

```python
# Basic information after cleaning
print("Dataset shape:", df_clean.shape)

print("\nMissing values:")
print(df_clean.isnull().sum())

print("\nDuplicate rows:")
print(df_clean.duplicated().sum())

print("\nFirst 5 rows:")
display(df_clean.head())
```

```python
# Step 3.2 - Statistical Summary

summary = df_clean[numeric_columns].agg(
    ['mean', 'median', 'min', 'max']
)

print("Mean, Median, Minimum and Maximum:")
print(summary)
```

```python
# Gender distribution
gender_counts = df_clean["Gender"].value_counts()

print("Gender Distribution:")
display(gender_counts)

gender_percentage = df_clean["Gender"].value_counts(normalize=True)*100
print("Genter percentage")
display(gender_percentage)
```

```python
# Discriptive Analytics
```

```python
print("DESCRIPTIVE ANALYTICS")
print("=" * 50)

# Number of users
print("Total number of users:", df_clean["Users"].nunique())

# Average values
print("\nAverage values:")
print("Average Posts:", df_clean["Number of Posts"].mean())
print("Average Followers:", df_clean["Number of Followers"].mean())
print("Average Followings:", df_clean["Number of Followings"].mean())
```

```python
'''Descriptive Statistics for Likes
My dataset has likes from the 10th, 11th and 12th previous posts.'''
print("Average Likes:")
print(
    "10th Previous Post:",
    df_clean["Number of Likes for the Tenth Previous Post"].mean()
)

print(
    "11th Previous Post:",
    df_clean["Number of Likes for the Eleventh Previous Post"].mean()
)

print(
    "12th Previous Post:",
    df_clean["Number of Likes for the Twelfth Previous Post"].mean()
)
```

```python
# Minimum and maximum values

descriptive_table = df_clean[numeric_columns].agg(
    ["min", "max", "mean", "median"]
)

print("Minimum, Maximum, Mean and Median:")
print(descriptive_table)
```

```python
 # Descriptive analysis by gender

gender_analysis = df_clean.groupby("Gender")[numeric_columns].mean()

print("Average values by Gender:")
print(gender_analysis)
```

```python
# Average Likes by Gender

gender_likes = df_clean.groupby("Gender")[
    [
        "Number of Likes for the Tenth Previous Post",
        "Number of Likes for the Eleventh Previous Post",
        "Number of Likes for the Twelfth Previous Post"
    ]
].mean()

print("Average Likes by Gender:")
print(gender_likes)
```

```python
# Final Descriptive Statistics Table
descriptive_summary = pd.DataFrame({
    "Metric": [
        "Number of Users",
        "Average Posts",
        "Average Followers",
        "Average Followings",
        "Average Likes - 10th Post",
        "Average Likes - 11th Post",
        "Average Likes - 12th Post"
    ],
    "Value": [
        df_clean["Users"].nunique(),
        df_clean["Number of Posts"].mean(),
        df_clean["Number of Followers"].mean(),
        df_clean["Number of Followings"].mean(),
        df_clean["Number of Likes for the Tenth Previous Post"].mean(),
        df_clean["Number of Likes for the Eleventh Previous Post"].mean(),
        df_clean["Number of Likes for the Twelfth Previous Post"].mean()
    ]
})
print(descriptive_summary)
```

```python
'''What is Diagnostic Analytics?
Descriptive analytics asks:
What happened?
Diagnostic analytics asks:
Why might it have happened?'''
# 1 Does more followers means more likes

followers_likes = df_clean[
    [
        "Number of Followers",
        "Number of Likes for the Tenth Previous Post"
    ]
].corr()

print("Correlation between Followers and Likes:")
print(followers_likes)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 5))
plt.scatter(
    df_clean["Number of Followers"],
    df_clean["Number of Likes for the Tenth Previous Post"]
)
plt.title("Followers vs Likes")
plt.xlabel("Number of Followers")
plt.ylabel("Likes on Tenth Previous Post")
plt.show()
'''What does your graph show?
Looking at your graph:
Most users are concentrated toward the lower-left → they have relatively fewer followers and fewer likes.
As the number of followers increases, the likes generally tend to increase.
There are some users with very high followers and very high likes.
However, the points are quite spread out, so the relationship isn't perfectly strong.
There are also some outliers — users whose follower count or likes are much higher than most users.
So, visually, we can say:
There appears to be a positive relationship between the number of followers and the number of likes. Users with more followers tend to receive more likes, although the relationship is not perfect.'''
```

```python
correlation_value = df_clean[
    [
        "Number of Followers",
        "Number of Likes for the Tenth Previous Post"
    ]
].corr().iloc[0, 1]

print("Correlation between Followers and Likes:", correlation_value)
```
