```python
import pandas as pd
import math
from scipy.stats import shapiro
df = pd.read_excel(r"C:\Users\aatif\OneDrive\Desktop\student_teaching_methods.xlsx")
print("First 5 rows")
print(df.head())
print("Last 5 rows")
print(df.tail())
```

```python
print("mean before")
mean_before = df["Marks_Before"].mean()
print(mean_before)
print("mean after")
mean_after = df["Marks_After"].mean()
print(mean_after)
```

```python
print("standers_deviation before")
stander_deviation_before = df["Marks_Before"].std()
print(stander_deviation_before)
print("standard deviation after")
stander_deviation_after = df["Marks_After"].std()
print(stander_deviation_after)
```

```python
mean_after = df["Marks_After"].mean()
std_after = df["Marks_After"].std()
n = len(df["Marks_After"])
t = mean_after / (std_after / math.sqrt(n))
print(t)
if t > 65:
    print("we will reject the null hypothesis")
else:
    print("we will not reject the null hypothesis")
```

```python
from scipy.stats import ttest_ind, levene

traditional = df[df["Teaching_Method"] == "Traditional"]["Marks_After"]
interactive = df[df["Teaching_Method"] == "Interactive"]["Marks_After"]

# 1. Normality test
print("Normality Test")

print("Traditional:", shapiro(traditional))
print("Interactive:", shapiro(interactive))

# 2. Equality of variances
print("\nLevene's Test")

levene_test = levene(traditional, interactive)
print(levene_test)

# 3. Independent samples t-test
print("\nIndependent Samples t-Test")

t_stat, p_value = ttest_ind(
    traditional,
    interactive,
    equal_var=True
)

print("t-statistic:", t_stat)
print("p-value:", p_value)

# 4. Decision
if p_value < 0.05:
    print("Reject H0: Significant difference")
else:
    print("Fail to reject H0: No significant difference")
```

```python
from scipy.stats import chi2_contingency
table = pd.crosstab(df["Teaching_Method"], df["Performance"])
print(table)
chi2, p_value, dof, expected = chi2_contingency(table)
print("expected frequencies")
print(expected)
expected_table = pd.DataFrame(
    expected,
    index=table.index,
    columns=table.columns
)

print(expected_table)
print("Chi-square statistic:", chi2)
print("Degrees of freedom:", dof)
print("p-value:", p_value)

print("Minimum expected frequency:", expected.min())
print("Expected frequencies:")
print(expected)

df["Performance_Group"] = df["Performance"].replace({
    "Average": "Below Good",
    "Poor": "Below Good",
    "Good": "Good or Above",
    "Excellent": "Good or Above"
})

table2 = pd.crosstab(
    df["Teaching_Method"],
    df["Performance_Group"]
)

print(table2)

chi2, p_value, dof, expected = chi2_contingency(table2)

print("Chi-square:", chi2)
print("Degrees of freedom:", dof)
print("p-value:", p_value)

print("Expected frequencies:")
print(expected)
```
