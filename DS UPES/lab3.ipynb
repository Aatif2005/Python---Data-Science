```python
# ONE-SAMPLE T-TEST
# Objective: Test whether the mean Marks_After is different
# from 65.
import numpy as np
import pandas as pd
from scipy.stats import t
marks_after = [
    60, 65, 52, 72, 61, 68, 48, 56, 63, 54,
    68, 75, 65, 82, 72, 78, 60, 69, 76, 66
]

# Convert data into NumPy array
marks = np.array(marks_after)

# Hypothesized population mean
mu = 65

# Significance level
alpha = 0.05

print(" ORIGINAL DATA")
print(marks)

# STEP 2: CALCULATE SAMPLE SIZE (N)
# Formula: N = Total number of observations
n = len(marks)

print("\n: SAMPLE SIZE")
print("Sample Size (N) =", n)

# STEP 3: CALCULATE SAMPLE MEAN
# Formula: X-bar = Sum(X) / N

total = np.sum(marks)

mean = total / n

print("\n SAMPLE MEAN")

print("Sum of Marks =", total)
print("Mean = Sum(X) / N")
print("Mean =", total, "/", n)
print("Sample Mean =", mean)

# STEP 4: CALCULATE DEVIATION FROM THE MEAN
# Formula: Deviation = Xi - X-bar

deviations = marks - mean

print("\n DEVIATION FROM MEAN")

deviation_table = pd.DataFrame({
    "Marks_After": marks,
    "Deviation (Xi - Mean)": deviations
})

print(deviation_table)

# STEP 5: CALCULATE SQUARED DEVIATIONS
# Formula: (Xi - X-bar)^2

squared_deviations = deviations ** 2

print("\nSQUARED DEVIATIONS")

squared_table = pd.DataFrame({
    "Marks_After": marks,
    "Deviation": deviations,
    "Squared Deviation": squared_deviations
})

print(squared_table)

sum_squared_deviations = np.sum(squared_deviations)

print("\nSum of Squared Deviations =",
      sum_squared_deviations)

# STEP 6: CALCULATE SAMPLE VARIANCE
# Formula: s^2 = Sum(Xi - X-bar)^2 / (N - 1)

variance = sum_squared_deviations / (n - 1)

print("\nSAMPLE VARIANCE")

print("Variance = Sum of Squared Deviations / (N - 1)")
print("Variance =", sum_squared_deviations, "/", n - 1)
print("Sample Variance =", variance)

# STEP 7: CALCULATE STANDARD DEVIATION
# Formula: s = Square Root of Variance

std_dev = np.sqrt(variance)

print("\n STANDARD DEVIATION")
print("Standard Deviation = sqrt(Variance)")
print("Standard Deviation =", std_dev)

# STEP 8: CALCULATE STANDARD ERROR
# Formula: SE = s / sqrt(N)

standard_error = std_dev / np.sqrt(n)

print("\nSTANDARD ERROR")
print("Standard Error = SD / sqrt(N)")
print("Standard Error =", standard_error)

# STEP 9: CALCULATE DEGREES OF FREEDOM
# Formula: df = N - 1

df = n - 1

print("\nDEGREES OF FREEDOM")
print("df = N - 1")
print("Degrees of Freedom =", df)

# STEP 10: CALCULATE T-STATISTIC
# Formula: t = (X-bar - mu) / (s / sqrt(N))

t_statistic = (mean - mu) / standard_error

print("\n T-STATISTIC")
print("t = (Sample Mean - Hypothesized Mean) / SE")
print("t =", mean, "-", mu, "/", standard_error)
print("T-Statistic =", t_statistic)

# STEP 11: CALCULATE TWO-TAILED P-VALUE
# Formula: p = 2 * [1 - CDF(|t|)]

p_value = 2 * (1 - t.cdf(abs(t_statistic), df)) # we have multiplied with 2 bcz oubles that, because the distribution is symmetric. There’s an equal-sized area in the lower tail (values below -|t|). Adding both tails gives the probability of a result at least this extreme in either direction

print("\n P-VALUE")
print("P-Value =", p_value)

# STEP 12: CALCULATE CRITICAL T-VALUE
# Two-tailed test with alpha = 0.05
# Formula: t_critical = t.ppf(1 - alpha/2, df)

critical_t = t.ppf(1 - alpha / 2, df)

print("\n CRITICAL T-VALUE")
print("Critical T-Value =", critical_t)

# STEP 13: HYPOTHESIS TESTING AND DECISION

print("\n HYPOTHESIS TESTING")

print("Null Hypothesis (H0): Mean = 65")
print("Alternative Hypothesis (H1): Mean != 65")

print("\nSignificance Level =", alpha)
print("Calculated P-Value =", p_value)

if p_value < alpha:

    print("\nDECISION: Reject Null Hypothesis")
    print("There is a statistically significant difference")
    print("between the sample mean and 65.")

else:

    print("\nDECISION: Fail to Reject Null Hypothesis")
    print("There is no statistically significant evidence")
    print("that the population mean differs from 65.")
```
